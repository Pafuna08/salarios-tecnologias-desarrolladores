Este cuaderno confirma que Jupyter usa el Python del entorno `.venv` y que pandas está disponible. También practica `if` / `elif` / `else` con una función que clasifica los **años de experiencia** de un desarrollador, una variable clave del proyecto.

```python
import sys
import pandas as pd

print(sys.version_info.major, sys.version_info.minor)
print(pd.__version__)
```

```python
def nivel_experiencia(anios_experiencia):
    """Clasifica los años de experiencia profesional en junior, semi-senior o senior."""
    if anios_experiencia < 3:
        return "junior"
    elif anios_experiencia < 8:
        return "semi-senior"
    else:
        return "senior"


for anios in [1, 5, 12]:
    print(anios, "->", nivel_experiencia(anios))
```
